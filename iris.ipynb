```python
import cv2

def print_image_information(image: cv2.MatLike):
    height, width, channels = image.shape

    print("Height   :", height)
    print("Width    :", width)
    print("Channels :", channels)
    print("Size     :", image.size / 1000, "KB") # h x w x c
    print("Data type:", image.dtype)

iris = cv2.imread("iris-1.jpg")

print_image_information(iris)
```
